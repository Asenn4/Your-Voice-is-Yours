# 🎤 AI Song Cover Studio - Notebook Edition
Notebook ini memungkinkan Anda untuk menjalankan RVC inference tanpa UI Streamlit secara langsung melalui kode.

```python
import os
import inference_wrapper

# Pastikan direktori input, output, dan models tersedia
os.makedirs("input", exist_ok=True)
os.makedirs("output", exist_ok=True)
os.makedirs("models", exist_ok=True)
```

## 1. Memuat Model RVC
Tentukan lokasi file `.pth` (dan opsional `.index`). Model yang sebelumnya Anda tambahkan via Streamlit akan tersimpan di dalam folder `models/`.

```python
# Ganti dengan nama folder dan file model Anda
MODEL_PATH = "models/NAMA_MODEL_ANDA/file_model.pth"
INDEX_PATH = "" # (Opsional) Isi path file .index jika ada, misalnya: "models/NAMA_MODEL_ANDA/file_index.index"

if os.path.exists(MODEL_PATH):
    print(f"Memuat model dari: {MODEL_PATH}")
    inference_wrapper.load_model(MODEL_PATH, index_path=INDEX_PATH)
    print("✅ Model berhasil dimuat!")
else:
    print(f"❌ Model tidak ditemukan di path: {MODEL_PATH}")
```

## 2. Proses Konversi (Inference)
Tentukan file vokal asli (`.wav` atau `.mp3`) yang sudah dimasukkan ke folder `input/`, lalu jalankan sel di bawah ini.

```python
INPUT_AUDIO = "input/vokal_asli.wav" # Ganti dengan nama file audio input Anda
OUTPUT_AUDIO = "output/hasil_cover.wav"

# Atur pitch (semitones): 
# 0 = suara asli, +12 = naik 1 oktaf (biasanya cowok ke cewek), -12 = turun 1 oktaf (cewek ke cowok)
PITCH = 0 

if not os.path.exists(INPUT_AUDIO):
    print(f"❌ File input tidak ditemukan: {INPUT_AUDIO}")
else:
    print(f"🚀 Memproses {INPUT_AUDIO} menjadi {OUTPUT_AUDIO}...")
    success, err_msg = inference_wrapper.convert_audio(INPUT_AUDIO, OUTPUT_AUDIO, pitch=PITCH)
    
    if success:
        print(f"\n🎉 Sukses! Hasil disimpan di: {OUTPUT_AUDIO}")
    else:
        print(f"\n❌ Gagal memproses audio. Error:\n{err_msg}")
```

## 3. Putar Hasil Audio

```python
from IPython.display import Audio, display

if os.path.exists(OUTPUT_AUDIO):
    display(Audio(OUTPUT_AUDIO))
else:
    print("Belum ada hasil audio yang dapat diputar.")
```
